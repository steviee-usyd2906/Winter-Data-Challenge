```python
import galah
import pandas as pd

# 1. Configure ALA API (You must register this email at ala.org.au first)
galah.galah_config(email="hieudonguyenthanh2906@gmail.com")

# 2. Define the states to pull data for
states_to_pull = [
    "New South Wales",
    "Victoria",
    "Queensland"
]

# 3. Define your list of canopy-dependent indicator species
# These species rely on mature forest structure (hollows/canopy) which takes decades to recover.
indicator_species = [
    "Phascolarctos cinereus",  # Koala
    "Petauroides volans",      # Greater Glider
    "Petaurus australis"       # Yellow-bellied Glider
]

# 4. Create an empty list to store the dataframes
all_sightings = []

# 5. Loop through each state, download the multi-species data, and append it
print("Downloading multi-species ALA data for NSW, VIC, and QLD...")
for state in states_to_pull:
    print(f"Pulling data for {state}...")
    
    # Pass the entire list of species to the 'taxa' parameter
    sightings_df = galah.atlas_occurrences(
        taxa=indicator_species,
        filters=[
            "year>=1990", 
            "year<=2026", 
            f"stateProvince={state}"
        ]
    )
    
    all_sightings.append(sightings_df)

# 6. Combine all the state dataframes into one single dataframe
combined_sightings_df = pd.concat(all_sightings, ignore_index=True)

# 7. Save the combined data to a CSV file for later use
output_filename = "canopy_indicators_nsw_vic_qld.csv"
combined_sightings_df.to_csv(output_filename, index=False)

print(f"Success! Data saved locally as '{output_filename}'.")
```

```python
import pandas as pd
import plotly.express as px

# 1. Load the CSV you saved in the previous step
output_filename = "canopy_indicators_nsw_vic_qld.csv"
df = pd.read_csv(output_filename)

# 2. Drop any rows that are missing GPS coordinates (cleans the data for plotting)
df = df.dropna(subset=['decimalLatitude', 'decimalLongitude'])

# 3. Create the interactive map
print("Generating map... this might take a moment depending on the data size.")
fig = px.scatter_mapbox(
    df, 
    lat="decimalLatitude", 
    lon="decimalLongitude", 
    color="scientificName",        # This will give each of your 3 species a different color!
    hover_name="scientificName",   # Shows the species name when you hover your mouse
    hover_data=["eventDate"],      # Shows the date of the sighting on hover
    zoom=4,                        # Sets the initial zoom level to view the East Coast
    center={"lat": -28.0, "lon": 147.0}, # Centers the map generally over NSW/QLD/VIC
    mapbox_style="open-street-map", 
    title="Canopy-Dependent Indicator Species Sightings (1990 - 2026)",
    opacity=0.6                    # Makes the dots slightly transparent to see dense clusters
)

# 4. Display the map in your browser or Jupyter Notebook
fig.show(renderer="browser")
```
